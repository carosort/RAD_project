```python
import urllib.request

def fetch_89bit_int():
    # Hent 12 bytes = 96 bits
    url = "https://www.random.org/cgi-bin/randbyte?nbytes=12&format=b"
    data = urllib.request.urlopen(url).read()

    # Konverter bytes → heltal (96 bits)
    full_int = int.from_bytes(data, byteorder="big")

    # Maskér så kun de første 89 bits bevares
    # (dvs. fjern de nederste 7 bits)
    top_89_bits = full_int >> 7

    return top_89_bits


def generate_sequences():
    for _ in range(100):
        values = [fetch_89bit_int() for _ in range(4)]
        print(f"[|{values[0]}I;{values[1]}I;{values[2]}I;{values[3]}I|]")


# Kør
generate_sequences()
```

```python
import requests

def fetch_89_bits_decimal():
    # Hent 12 bytes (96 bits) som en binær streng med mellemrum mellem bytes
    url = "https://www.random.org/cgi-bin/randbyte?nbytes=12&format=b"
    raw = requests.get(url).text.strip()

    # Fjern mellemrum så vi får en ren bitstreng
    bitstring = raw.replace(" ", "")

    # Tag de første 89 bits
    first_89 = bitstring[:89]

    # Konverter til decimal
    decimal_value = int(first_89, 2)

    print("Første 89 bits:", first_89)
    print("Decimal:", decimal_value)

    return decimal_value

fetch_89_bits_decimal()
```
